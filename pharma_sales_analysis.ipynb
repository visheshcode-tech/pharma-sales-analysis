```python
df["Quantity_Sold"] = pd.to_numeric(
    df["Quantity_Sold"],
    errors="coerce"
)
```

```python
df["Unit_Price"] = pd.to_numeric(
    df["Unit_Price"],
    errors="coerce"
)
```

```python
df["City"] = df["City"].fillna("Unknown")
df["Quantity_Sold"] = df["Quantity_Sold"].fillna(
    df["Quantity_Sold"].median()
)
df["Unit_Price"] = df["Unit_Price"].fillna(df["Unit_Price"].median())

df["Sales_Rep"] = df["Sales_Rep"].fillna("Unknown")
df["Discount_Percent"] = df["Discount_Percent"].fillna("Unknown")


```

```python
df.head(23)
```

```python
df["Discount_Percent"] = pd.to_numeric(df["Discount_Percent"], errors="coerce")
df["Discount_Percent"] = df["Discount_Percent"].fillna(0)
```

```python
df["Sale_Date"] = pd.to_datetime(df["Sale_Date"], format = "mixed" , dayfirst= True )
```

```python
df["Sale_Date"].head(34)
```

```python
df["Year"] = df["Sale_Date"].dt.year
df["Month"] = df["Sale_Date"].dt.month
```

```python
df["Gross_Sale"] = df["Quantity_Sold"]*df["Unit_Price"]
```

```python
df.head(34)
```

```python
df["Gross_Sale"].mean()
```

```python
df["Discount_Amount"] = (df["Gross_Sale"]*df["Discount_Percent"]/100)
```

```python
df.isnull().sum()
```

```python
df["Net_Sales"] = df["Gross_Sale"] - df["Discount_Amount"]
```

```python
df[["Gross_Sale", "Discount_Amount", "Net_Sales"]].head()
```

```python
Total_Sales = df["Net_Sales"].sum()
```

```python
print("total net sales :", Total_Sales )
```

```python
total_quantity = df["Quantity_Sold"].sum()

print("Total Quantity Sold:", total_quantity)
```

```python
average_sale = df["Net_Sales"].mean()

print("Average Sale:", average_sale)
```

```python
category_sales = df.groupby("Category")["Net_Sales"].sum()

print(category_sales)
```

```python
top_category = category_sales.idxmax()

print("Highest Selling Category:", top_category)
```

```python
category_sales.sort_values(ascending=False)
```

```python
region_sales = df.groupby("Region")["Net_Sales"].sum()

region_sales
```

```python
top_region = region_sales.idxmax()

print("Highest Selling Region:", top_region)
```

```python
region_sales.plot(kind="bar", figsize=(8, 5) , color= "red")

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Net Sales")

plt.show()
```
